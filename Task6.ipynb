```python


#1. بوابة مواقف بتفتح لوحدها لو الشارع فاضي (حساس بيقيس المسافة).
#2. تطبيق طلبات بيختار أقرب مندوب للعميل الجديد بالتعلّم من آلاف الطلبات القديمة.
#3. حاسبة سوبر ماركت: الإجمالي = مجموع الأسعار + ضريبة 14%.
#4. تطبيق يترجم صوتك من لغة لغة تانية فورًا.
#5. لمبة بتنور لو حركة اتلمست قدامها.
#6. موقع إخباري بيقترحلك مقالات بيشبه اللي قريته في آخر أسبوع.
#7. توقّع لو حد هيربح اليانصيب الجاي.
#8. برنامج يحلّل رسائلك الشخصية بآلة بدون إذنك.
#9. تصفية رسائل السبام اللي بتيجي على إيميلك.
#10. تذكير تلقائي بموعد الاشتراك الشهري كل 28 من الشهر.

answers = {
    "1": "قواعد", "2": "تعلّم", "3": "قواعد", "4": "تعلّم", "5": "قواعد",
    "6": "تعلّم", "7": "مش مناسب", "8": "تعلّم", "9": "تعلّم", "10": "قواعد",
}
# TODO: اكتب لكل مفتاح واحد من: "قواعد" | "تعلّم" | "مش مناسب"
choices = {"قواعد", "تعلّم", "مش مناسب"}
assert len(answers) == 10, "عشرة مواقف"
assert all(v in choices for v in answers.values()), "كل اختيار من الأقسام الثلاثة"
print("A1: ممتاز")
```
