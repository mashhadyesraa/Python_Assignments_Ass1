```python
cart = []
cart.append("Latte")
cart.append("Croissant")
cart.append("Tea")
cart.remove("Croissant")  # شيل الأول ليه قيمته
cart.insert(0, "Water")  # ضيف في الأول
cart.pop()               # شيل آخر عنصر (Tea)
cart.sort()              # رتب داخل نفس القائمة
print(cart, len(cart))
assert cart == ["Latte", "Water"], "شوف ترتيب الشغل: append/remove/insert/pop/sort"
assert len(cart) == 2
print("A1: ممتاز")
```

```python
grades = {"math": 88, "english": 76}

# TODO: تعديل نفس المفتاح، قيمة جديدة #
grades["math"] = 92

# TODO: مفتاح جديد = إضافة #
grades["physics"] = 79

# TODO: حذف english #
del grades["english"]

# TODO: أعلى درجة باستخدام max() دالة جاهزة في بايثون #
best = max(grades.values())

print(grades, best)

# تحقق ذاتي:
assert grades == {"math": 92, "physics": 79}
assert best == 92
print("A3: ممتاز")
```
