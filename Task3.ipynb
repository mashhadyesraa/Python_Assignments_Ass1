```python
grade = 95
level = None

if grade >= 90:
    level = "A"
elif grade >= 80:
    level = "B"
elif grade >= 70:
    level = "C"
elif grade >= 60:
    level = "D"
else:
    level = "F"

print(f"الدرجة {grade} -> {level}")

# التحقق الذاتي (انسخه تحت الكود)
assert level in ("A", "B", "C", "D", "F"), "اكتب شرط if/elif كامل الأول"
assert not (grade >= 90 and level != "A"), ">=90: A"
assert not (80 <= grade < 90 and level != "B"), "80-89: B"
assert not (70 <= grade < 80 and level != "C"), "70-79: C"
assert not (60 <= grade < 70 and level != "D"), "60-69: D"
assert not (grade < 60 and level != "F"), "<60: F"

print("A1: ممتاز")
```

```python
import random

target = random.randint(1, 50)
tries = 0

guess = int(input("خمن رقم من 1 لـ 50: "))
tries += 1

while guess != target:
    if guess < target:
        print("أكثر")
    else:
        print("أقل")
    
    guess = int(input("خمن تاني: "))
    tries += 1

print(f"برافو! عدد المحاولات: {tries}")

assert tries > 0, "العب اللعبة الأول"
print("A3: ممتاز")
```

```python
scores = [82, 90, 76, 88, 95]
total = 0
count = 0
biggest = scores[0]

for s in scores:
    total += s
    count += 1
    if s > biggest:
        biggest = s

average = total / count
print(total, round(average, 2), biggest)

assert total == 431, "total = مجموع الخمس درجات"
assert count == 5, "count بيزود مرة في كل دور"
assert biggest == 95, "biggest يتغير بس لو لقينا أكبر"
assert abs(average - 86.2) < 0.001, "average = total / count"

print("A4: ممتاز")
```

```python
FIELDS = ("meal_id", "meal_name", "preparation_time", "calories", "meal_category", "day")

records = [
    {"meal_id": 201, "meal_name": "Grilled Chicken Salad", "preparation_time": 25, "calories": 450.5, "meal_category": "Lunch", "day": "Friday"},
    {"meal_id": 202, "meal_name": "Oatmeal with Berries", "preparation_time": 10, "calories": 300.0, "meal_category": "Breakfast", "day": "Saturday"},
    {"meal_id": 203, "meal_name": "Beef Burger Meal", "preparation_time": 30, "calories": 850.0, "meal_category": "Lunch", "day": "Sunday"},
    {"meal_id": 204, "meal_name": "Greek Yogurt Parfait", "preparation_time": 5, "calories": 220.0, "meal_category": "Snack", "day": "Monday"},
    {"meal_id": 205, "meal_name": "Tuna Sandwich", "preparation_time": 15, "calories": 380.0, "meal_category": "Breakfast", "day": "Tuesday"},
    {"meal_id": 206, "meal_name": "Salmon with Quinoa", "preparation_time": 35, "calories": 520.0, "meal_category": "Lunch", "day": "Wednesday"},
    {"meal_id": 207, "meal_name": "Protein Shake", "preparation_time": 5, "calories": 250.0, "meal_category": "Snack", "day": "Thursday"},
    {"meal_id": 208, "meal_name": "Pasta Alfredo", "preparation_time": 25, "calories": 700.0, "meal_category": "Lunch", "day": "Friday"},
    {"meal_id": 209, "meal_name": "Avocado Toast", "preparation_time": 10, "calories": 320.0, "meal_category": "Breakfast", "day": "Saturday"},
    {"meal_id": 210, "meal_name": "Fruit Salad", "preparation_time": 10, "calories": 180.0, "meal_category": "Snack", "day": "Sunday"}
]

while True:
    print("\n--- القائمة الرئيسية ---")
    print("1. عرض كل السجلات")
    print("2. إضافة سجل")
    print("3. تصفية حسب التصنيف (meal_category)")
    print("4. ملخص (عدد، مجموع، متوسط، أعلى، أدنى للسعرات calories)")
    print("5. بحث بالكلمة")
    print("6. خروج")
    
    choice = input("اختار رقم من القائمة (1-6): ").strip()
    
    if choice == "1":
        print("\n--- كل السجلات ---")
        for r in records:
            print(r)
            
    elif choice == "2":
            print("\n--- إضافة وجبة جديدة (تلقائية الـ ID) ---")
            try:
                if records:
                    new_id = max(r["meal_id"] for r in records) + 1
                else:
                    new_id = 201  # لو القائمة فارغة نبدأ من 201
                
                print(f"تم تخصيص معرف تلقائي للوجبة: {new_id}")
                
                m_name = input("أدخل اسم الوجبة (meal_name): ")
                m_time = int(input("وقت التحضير preparation_time (رقم): "))
                m_cal = float(input("السعرات calories (رقم): "))
                m_cat = input("التصنيف meal_category (مثال: Lunch, Breakfast...): ")
                m_day = input("اليوم day: ")
                
                new_record = {
                    "meal_id": new_id,
                    "meal_name": m_name,
                    "preparation_time": m_time,
                    "calories": m_cal,
                    "meal_category": m_cat,
                    "day": m_day
                }
                records.append(new_record)
                print("تم إضافة الوجبة بنجاح!")
            except ValueError:
                print("خطأ: يرجى إدخال قيم صحيحة للحقول الرقمية (الوقت أو السعرات). تم إلغاء الإضافة.")         
    elif choice == "3":
        target_cat = input("أدخل التصنيف المطلوب للبحث عنه (مثل Lunch أو Breakfast): ").strip().capitalize()
        print(f"\n--- سجلات التصنيف: {target_cat} ---")
        found = False
        for r in records:
            if r["meal_category"].capitalize() == target_cat:
                print(r)
                found = True
        if not found:
            print("لا توجد سجلات بهذا التصنيف.")
            
    elif choice == "4":
        # ملخص للسعرات calories (العدد، المجموع، المتوسط، الأعلى، الأدنى)
        if not records:
            print("لا توجد سجلات لحساب الملخص.")
        else:
            count = 0
            total_cal = 0.0
            highest_cal = records[0]["calories"]
            lowest_cal = records[0]["calories"]
            
            for r in records:
                cal = r["calories"]
                count += 1
                total_cal += cal
                if cal > highest_cal:
                    highest_cal = cal
                if cal < lowest_cal:
                    lowest_cal = cal
            
            avg_cal = total_cal / count if count > 0 else 0
            
            print("\n--- ملخص السعرات (Calories Summary) ---")
            print(f"العدد الإجمالي: {count}")
            print(f"مجموع السعرات: {total_cal}")
            print(f"متوسط السعرات: {round(avg_cal, 2)}")
            print(f"أعلى سعرات: {highest_cal}")
            print(f"أقل سعرات: {lowest_cal}")
            
    elif choice == "5":
        keyword = input("أدخل الكلمة للبحث عنها في كل الحقول: ").strip().lower()
        print(f"\n--- نتائج البحث عن: '{keyword}' ---")
        found = False
        for r in records:
            # تحويل قيم الدكشنري كلها لنصوص للبحث السهل
            match = any(keyword in str(val).lower() for val in r.values())
            if match:
                print(r)
                found = True
        if not found:
            print("لم يتم العثور على أي سجل يحتوي على هذه الكلمة.")
            
    elif choice == "6":
        print("مع السلامة! تم الخروج من المنيو بنجاح.")
        break
        
    else:
        print("اختيار غلط - جرب تاني (اختر من 1 لـ 6).")

# --- التحقق البرمجي النهائي (ألصقه في آخر النيوتبوك بعد الخروج) ---
assert isinstance(records, list), "records = list"
assert len(records) >= 10, "10 سجلات على الأقل"
assert all(isinstance(r, dict) for r in records), "كل سجل dict"
assert all("calories" in r for r in records), "حقل calories موجود"
assert all("meal_category" in r for r in records), "حقل meal_category موجود"
print("Part B: ممتاز")
```
